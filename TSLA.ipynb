```python
url = 'https://raw.githubusercontent.com/bprasad26/lwd/master/data/tesla_stock_prices.csv'
tesla_df = pd.read_csv(url, parse_dates=['Date'])      
tesla_df.head()
```

```python
tesla_df['month'] = tesla_df['Date'].dt.month
tesla_df['day_of_month  '] = tesla_df['Date'].dt.day
tesla_df['year'] = tesla_df['Date'].dt.year 

tesla_df.head()
```

```python
# Add new columns to dataframe based off of dates
tesla_df['Day'] = tesla_df['Date'].dt.day
tesla_df['Month'] = tesla_df['Date'].dt.month
tesla_df['Year'] = tesla_df['Date'].dt.year
tesla_df['day_name'] = tesla_df['Date'].dt.day_name()

tesla_df.head(10)
```

```python
least_recent_date = tesla_df['Date'].min()
most_recent_date= tesla_df[tesla_df['Date']== tesla_df['Date'].max()]['Date'].values[0]     
print(most_recent_date)
print(least_recent_date)
```

```python
monthly_sales = tesla_df.groupby(['month', 'year'])['Volume'].sum().reset_index()
monthly_sales.head(15)
```

```python
## Filtering using Loc or Query
filtered_df = tesla_df.loc[tesla_df['Date'] >= '2020-08-01']
filtered_df_date = tesla_df.query('Date >= "2020-08-01"')
filtered_df.head()  
```
