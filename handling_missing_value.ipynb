Handling MIssing Values
-   `df.isnull()` - to detect missing values (`.isna()` is the same thing)
-   `df.fillna()` - to fill missing values (imputation)
-   `df.dropna()` - to remove missing values
-   `df.replace()`-  to replace specific values (imputation)

```python
import numpy as np
import pandas as pd 
```

```python
employee_list = [['James', 36, 75, 5428000],
               ['Villers', 38, 74, 3428000],
               ['VKole', 31, 70, 8428000],
               ['Smith', 34, 80, 4428000],
               ['Gayle', 40, 100, 4528000],
               ['Adam', 40, np.nan, 4528000],
               ['Rooter', 33, 72, 7028000],
               ['Peterson', 42, 85, 2528000],
               ['lynda', 42, 85, np.nan],
               [np.nan, 42, 85, np.nan],
               ['Jenny', np.nan, 100, 25632],
               ['Kenn', np.nan, 110, 25632],
                ['Aly', np.nan, 90, 25582],
               ['John', 41, 85, 1528000]]

employee_df = pd.DataFrame(employee_list, columns=['Name', 'Age', 'Weight', 'Salary'])
```

```python
rows, columns = employee_df.shape
print(f'The number of rows is {rows} and the number of columns is {columns}')
```

```python
employee_df.head()
```

```python
employee_df.isnull().sum()
```

```python
# be weary of the missing values in the DataFrame. It can cause problems when we are doing data analysis and machine learning. 
dropped_employee_df = employee_df.dropna()
print(dropped_employee_df)  
```

```python
filled_employee_df = employee_df.fillna(0)
print(filled_employee_df)
```

```python
filled_employee_df.isnull().sum()
```

```python
employee_df2 = employee_df.fillna(value={'Weight': 'Pending', 'Age': 'Unknown', 'Salary': '0.0', 'Name': 'Verification Pending'})
print(employee_df2)
```

```python
technologies = {
    'Courses':["Spark",np.nan,"Hadoop","Python","pandas",np.nan,"Java"],
    'Fee' :[20000,25000, np.nan,22000,24000,np.nan,22000],
    'Duration':[np.nan,'40days','35days', np.nan,'60days','50days','55days'],
    'Discount':[1000,np.nan,1500,np.nan,2500,2100,np.nan]
              }

tech=pd.DataFrame(technologies)
print(tech)
```

```python
rows, columns = tech.shape
print(f'The number of rows is {rows} and the number of columns is {columns}')
```

```python
tech.isnull().sum()
```

```python
tech
```

```python
tech2=tech.replace(np.nan, '')
print(tech2)
```

```python
tech3=tech.replace({'Courses': np.nan, 'Fee': np.nan, 'Duration': np.nan, 'Discount': np.nan})
print(tech3)  
```
